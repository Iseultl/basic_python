```python
# Exercise 6
temps = {"Marta": 36.8, "Jordi": 38.4, "Aina": 37.2, "Pau": 39.1, "Laia": 36.5}

for name, temp in temps.items():
    if temp >= 38.0:
        print(name, temp, "fever")
    else:
        print(name, temp, "normal")

fever = []
for name, temp in temps.items():
    if temp >= 38.0:
        fever.append(name)
print(fever, len(fever))          # ['Jordi', 'Pau'] 2
```

```python
# Extra exercise
def dose(weight_kg, mg_per_kg=10):
    return weight_kg * mg_per_kg

weights = {"Marta": 60, "Jordi": 82, "Aina": 45}
for name, w in weights.items():
    d = dose(w)
    print(name, d, "mg")
    if d > 700:
        print("  check with the doctor")   # only Jordi, 820 mg
```
