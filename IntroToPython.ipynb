# Step 1: Variables and numbers
- A variable is a name that points to a value
- `int` holds whole numbers, `float` holds numbers with a decimal point
- `type()` tells you what kind of value a variable holds
- Lines starting with `#` are comments; Python ignores them

````python
# Create variables: no need to declare the type
cell_count = 150        # int
temperature = 36.6      # float

print("Cells:", cell_count, type(cell_count))
print("Temperature:", temperature, type(temperature))

# Basic arithmetic
a = 15
b = 4
print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)    # division always gives a float
print("a // b =", a // b)   # whole-number division
print("a ** 2 =", a ** 2)   # power
````

# Step 2: Strings
- A string is text inside quotes: `"ATGC"` or `'ATGC'`
- Each character has a position (index) that starts at **0**
- `text[start:end]` cuts out a piece; the `end` position is not included
- Strings come with built-in methods such as `.upper()`, `.count()` and `.replace()`

````python
dna = "atgcgatcgt"

# Indexing and slicing
print("First base:", dna[0])
print("Last base:", dna[-1])          # negative indices count from the end
print("First three bases:", dna[0:3])
print("Length:", len(dna))

# Useful methods
dna = dna.upper()
print("Uppercase:", dna)
print("Number of G:", dna.count("G"))
print("T replaced by U (RNA):", dna.replace("T", "U"))
print("Contains 'GAT'?", "GAT" in dna)
````

# Step 3: Lists
- A list holds several values in order, inside square brackets `[]`
- Indexing and slicing work exactly like they do for strings
- Unlike strings, lists can be changed after you create them
- `.append()` adds to the end, `.remove()` deletes a value, `len()` counts items

````python
shopping = ["apple", "banana", "chocolate"]
print(shopping)

# Access items by position
print("First item:", shopping[0])
print("Last item:", shopping[-1])
print("Items 1 and 2:", shopping[1:3])

# Change, add and remove items
shopping[0] = "cheese"          # replace
shopping.append("milk")         # add to the end
shopping.remove("banana")       # remove by value
print(shopping)
print("Items in the list:", len(shopping))
````

# Step 4: Dictionaries
- A dictionary stores **key: value** pairs inside curly brackets `{}`
- You look up a value with its key, not with a position
- Assigning to a new key adds it; assigning to an existing key overwrites it
- `.get()` is a safe lookup: it returns `None` instead of crashing when the key is missing

````python
calendar = {"Monday": "Gym", "Tuesday": "Bioinformatics"}
print(calendar)

# Look up values
print(calendar["Monday"])
print(calendar.get("Friday"))              # None: the key does not exist

# Add, update and delete
calendar["Wednesday"] = "Python"    # new key
calendar["Monday"] = "Running"      # existing key, so the value is replaced
del calendar["Tuesday"]
print(calendar)

# See all keys and values
print("Keys:", list(calendar.keys()))
print("Values:", list(calendar.values()))
````

# Step 6: For loops and conditions
- A `for` loop runs the same code once for every item in a collection
- The indented lines (4 spaces) belong to the loop
- `if`, `elif` and `else` choose which code to run depending on a condition
- Comparisons: `==` equal, `!=` not equal, `>` `<` `>=` `<=`, and combine them with `and` / `or`

````python
shopping = ["chocolate", "banana", "apple"]

# Loop over a list
for item in shopping:
    print(item.upper())

# Loop with a decision
for food in shopping:
    if food == "chocolate":
        print("Buy", food)
    elif food == "banana":
        print("Maybe buy", food)
    else:
        print("Skip", food)

# Loop over a dictionary: .items() gives you the key and the value together
calendar = {"Monday": "Gym", "Tuesday": "Bioinformatics", "Wednesday": "Python"}
for day, activity in calendar.items():
    print(f"On {day} I do {activity}")

# A common pattern: start a counter at 0 and add to it inside the loop
seqs = ["ATGCGT", "TTAGGC", "CCGTAA", "GG"]
long_seqs = 0
for s in seqs:
    if len(s) >= 6:
        long_seqs = long_seqs + 1
print("Sequences with 6 or more bases:", long_seqs)
````

# Extra: Functions (if time allows)
- A function is a named block of code you can run again and again
- Create one with `def`, give it inputs (parameters), and send a result back with `return`
- Parameters can have default values that are used when you leave them out

````python
# A function with one input and one output
def gc_content(sequence):
    sequence = sequence.upper()
    gc = sequence.count("G") + sequence.count("C")
    return gc / len(sequence) * 100

print(gc_content("ATGCGATCGA"))

# A parameter with a default value
def count_base(sequence, base="A"):
    return sequence.upper().count(base)

print(count_base("ATGCGAG"))        # counts A
print(count_base("ATGCGAG", "G"))   # counts G

# Use a function inside a loop
for s in ["ATATAT", "GCGCAT", "GGGGCC"]:
    print(s, gc_content(s))

# Watch out: without return, a function gives back None
def add_no_return(x, y):
    result = x + y

print(add_no_return(2, 3))   # None
````

# Step 5: Working with large amounts of data with groupby
- When we want summary statistics for different groups in our dataframe
- For example calculating the mean lifespan per taxonomic class

## Tip:
- GroupBy is very powerful in Pandas for summarizing data by categories.

````python
# Group by Class and calculate mean MaxLongevity
mean_lifespan = df_clean.groupby('Class')['Maximum longevity (yrs)'].mean()
print(f"\nMean lifespan per class: {mean_lifespan.sort_values(ascending=False)}")

species_per_class = df_clean.groupby('Class')['Common name'].nunique()
print(f"\nSpecies per class: {species_per_class}")

mean_body_mass = df_clean.groupby('Class')['Body mass (g)'].mean()
print(f"\nMean body mass per class: {mean_body_mass}")
````

# Step 6: Visualizing Data with Matplotlib
Matplotlib is a python library for creating plots and figures. It is highly customizable.

- Turns data into visual stories that you can see and understand.

## How Matplotlib thinks...
Builds in layers
1. Create a figure (the canvas)
2. Add axes (the plotting area)
3. Draw your data into it

````python
# import the matplotlib library
import matplotlib.pyplot as plt
````

````python
# Drop rows with missing MaxLongevity or BodyWeight
df_clean = df.dropna(subset=['Maximum longevity (yrs)', 'Body mass (g)'])

# Histogram of maximum lifespans
plt.figure(figsize=(10,6))

plt.hist(df_clean['Maximum longevity (yrs)'], bins=100, color='skyblue', edgecolor='black')

plt.title("Distribution of Maximum Longevity Across Species")
plt.xlabel("Max Longevity (years)")
plt.ylabel("Number of Species")
plt.show()
````

# Step 7: Scatter Plot
- Scatter plots allow us to compare values from two columns against each other.

````python
# Scatter Plot - Body Weight vs Max Longevity
plt.figure(figsize=(10,6))
plt.scatter(df_clean['Body mass (g)'], df_clean['Maximum longevity (yrs)'], alpha=0.2, color='green')
plt.title("Body Weight vs Max Longevity")
plt.xlabel("Body Weight (kg)")
plt.ylabel("Max Longevity (years)")
plt.show()
````

# Improving our figures
- Adding annotations with colour and labelling certain datapoints makes our figures more informative
- Visualising points on the log scale can help by separating our data and making it more understandable

````python
# Step 8.2: Scatter Plot - colored by Order and label longest-lived species
plt.figure(figsize=(12,8))

# Get unique orders and assign colors <- Colour each point by the phylogenetic tree
orders = df_clean['Order'].unique()
colors = plt.cm.tab20(np.linspace(0, 1, len(orders)))
order_color_map = dict(zip(orders, colors))

# Plot each point with color based on Order
for idx, row in df_clean.iterrows():
    plt.scatter(row['Body mass (g)'], row['Maximum longevity (yrs)'],
                color=order_color_map[row['Order']], alpha=0.6)

# Label the top 5 longest-lived species
top_longest = df_clean.nlargest(10, 'Maximum longevity (yrs)')
for idx, row in top_longest.iterrows():
    plt.text(row['Body mass (g)'], row['Maximum longevity (yrs)'],
             row['Common name'], fontsize=9, ha='right', va='bottom')

# Plot the layers for the Matplotlib figure - Title & Axis
plt.xscale('log')  # log scale for better visualization
plt.title("Body Weight vs Max Longevity")
plt.xlabel("Body Weight (g, log scale)")
plt.ylabel("Maximum Longevity (yrs)")

# Create a legend for Orders
from matplotlib.patches import Patch
legend_elements = [Patch(facecolor=color, label=order) for order, color in order_color_map.items()]
plt.legend(handles=legend_elements, bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=8)

plt.show()
````
